```python
import numpy as np
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.preprocessing import LabelEncoder
```

### 1. Load Data
First, we'll load the student performance dataset from a CSV file into a pandas DataFrame. This is the initial step for any data analysis task, bringing our data into a structured format for manipulation.

```python
data = pd.read_csv("/content/student-mat - student-mat.csv", sep=',')
```

```python
print(data)
```

### 2. Feature Engineering: Create 'Pass' Column
We'll create a new target variable called 'Pass' based on the final grade ('G3'). If 'G3' is 10 or greater, the student 'Pass'es (1); otherwise, they 'Fail' (0). This transforms the problem into a binary classification task.

```python
# Convert final grade into Pass/Fail
data['Pass'] = np.where(data['G3'] >= 10, 1, 0)
```

```python
print(data)
```

### 3. Data Cleaning: Drop Original Grade Columns
Since we've created the 'Pass' column as our target variable, the original grade columns ('G1', 'G2', 'G3') are no longer needed for training and could lead to data leakage if used directly. We'll remove them from our dataset.

```python
# Drop original grade columns
data = data.drop(['G1', 'G2', 'G3'], axis=1)
```

```python
print(data)
```

### 4. Encode Categorical Variables
Many machine learning algorithms require numerical input. We'll convert categorical columns (e.g., 'school', 'sex', 'address') into numerical format using one-hot encoding with `pd.get_dummies`. The `drop_first=True` argument helps avoid multicollinearity.

```python
# Encode categorical variables
data = pd.get_dummies(data, drop_first=True)
```

```python
print(data)
```

### 5. Define Features (X) and Target (y)
Now, we'll separate our dataset into features (X), which are the input variables used to predict the outcome, and the target variable (y), which is what we want to predict ('Pass').

```python
# Define X and y
X = data.drop('Pass', axis=1)
y = data['Pass']
```

### 6. Split Data into Training and Testing Sets
To evaluate our models accurately, we'll split the data into training and testing sets. The training set is used to teach the model, and the testing set is used to assess its performance on unseen data. A `test_size` of 0.3 means 30% of the data will be used for testing, and `random_state` ensures reproducibility.

```python
# Split dataset
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)
```

### 7. Train and Evaluate Models for Underfitting, Balanced, and Overfitting Scenarios
We'll train three Decision Tree Classifier models with different `max_depth` parameters to demonstrate underfitting, balanced fitting, and overfitting.

*   **Underfitting Model (`max_depth=2`)**: A very simple model that might not capture enough patterns in the data.
*   **Balanced Model (`max_depth=5`)**: A moderately complex model aiming for a good balance between bias and variance.
*   **Overfitting Model (`max_depth=None`)**: A complex model that will fit the training data perfectly but might perform poorly on new data.

```python
# Underfitting Model
underfit_model = DecisionTreeClassifier(max_depth=2)

underfit_model.fit(X_train, y_train)

train_acc = accuracy_score(y_train, underfit_model.predict(X_train))
test_acc = accuracy_score(y_test, underfit_model.predict(X_test))

print("UNDERFITTING MODEL")
print(f"Train Accuracy: {train_acc:.3f}")
print(f"Test Accuracy: {test_acc:.3f}")
```

```python
# Balanced Model
balanced_model = DecisionTreeClassifier(max_depth=5)

balanced_model.fit(X_train, y_train)

train_acc = accuracy_score(y_train, balanced_model.predict(X_train))
test_acc = accuracy_score(y_test, balanced_model.predict(X_test))

print("BALANCED MODEL")
print(f"Train Accuracy: {train_acc:.3f}")
print(f"Test Accuracy: {test_acc:.3f}")
```

```python
# Overfitting Model
overfit_model = DecisionTreeClassifier(max_depth=None)

overfit_model.fit(X_train, y_train)

train_acc = accuracy_score(y_train, overfit_model.predict(X_train))
test_acc = accuracy_score(y_test, overfit_model.predict(X_test))

print("OVERFITTING MODEL")
print(f"Train Accuracy: {train_acc:.3f}")
print(f"Test Accuracy: {test_acc:.3f}")
```

### 8. Compare Model Performance
Finally, we'll compare the training and testing accuracies of all three models to observe the effects of underfitting, balanced fitting, and overfitting on performance. This helps us understand the trade-off between model complexity and generalization ability.

```python
# Compare performance
models = {
    "Underfitting": underfit_model,
    "Balanced": balanced_model,
    "Overfitting": overfit_model
}

for name, model in models.items():
    train_acc = accuracy_score(y_train, model.predict(X_train))
    test_acc = accuracy_score(y_test, model.predict(X_test))
    print(f"{name} → Train Accuracy: {train_acc:.3f}, Test Accuracy: {test_acc:.3f}")
```
